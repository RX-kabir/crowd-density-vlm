```python
import pandas as pd
import random
from pathlib import Path
```

```python
from pathlib import Path

PROJECT = Path(r"D:\CrowdDensityResearch")

TRAIN_CSV = PROJECT / "dataset" / "JHU-CROWD++" /"train"/ "train.csv"
TEST_CSV = PROJECT / "dataset" / "JHU-CROWD++" / "test"/"test.csv"

OUTPUT_FOLDER = PROJECT / "phase5" / "support_mapping"
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

OUTPUT_CSV = OUTPUT_FOLDER / "support_mapping.csv"
```

```python
print("Train CSV:", TRAIN_CSV.exists())
print("Test CSV :", TEST_CSV.exists())
```

```python
train_df = pd.read_csv(TRAIN_CSV)
test_df = pd.read_csv(TEST_CSV)

print("Training images :", len(train_df))
print("Testing images  :", len(test_df))
```

```python
SEED = 42
random.seed(SEED)
```

```python
train_lookup = {}

for _, row in train_df.iterrows():
    train_lookup[int(row["filename"])] = {
        "count": int(row["total_count"]),
        "density": int(row["crowd_density"])
    }

print("Lookup table created.")
print("Total training images:", len(train_lookup))
```

```python
mapping = []

train_ids = train_df["filename"].tolist()
```

```python
for _, test_row in test_df.iterrows():

    test_filename = int(test_row["filename"])

    supports = random.sample(train_ids, 5)

    row = {
        "test_filename": test_filename
    }

    for i, sid in enumerate(supports, start=1):

        sid = int(sid)

        row[f"support{i}_filename"] = sid
        row[f"support{i}_count"] = train_lookup[sid]["count"]
        row[f"support{i}_density"] = train_lookup[sid]["density"]

    mapping.append(row)
```

```python
mapping_df = pd.DataFrame(mapping)

mapping_df.head()
```

```python
mapping_df.to_csv(
    OUTPUT_CSV,
    index=False
)

print("Support mapping saved successfully!")
print(OUTPUT_CSV)
```
